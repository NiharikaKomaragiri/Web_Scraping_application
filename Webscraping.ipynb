```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
URL='https://timely-sunshine-e821b3.netlify.app/'
```

```python
page=requests.get(URL)
```

```python
page.status_code
```

```python
htmlCode=page.text
print(htmlCode)
```

```python
soup=BeautifulSoup(htmlCode,'html.parser')
print(soup)
```

```python
content=soup.find_all('div',attrs={'class':'a'})
for i in content:
  print("Div:",i,end="\n\n\n\n\n")
```

```python
content=soup.find('div',attrs={'class':'name'})
text=content.text.strip()
print(text)
```

```python
content=soup.find('div',attrs={'class':'price'})
text=content.text.strip()
print(text)
```

```python
content=soup.find('div',attrs={'class':'image'})
print(content)
```

```python
img_tag=content.find('img')
print(img_tag)
```

```python
img_src=img_tag['src']
print(img_src)
```

```python
items=soup.find_all('div',class_='a')

#lists to store extracted data
names=[]
prices=[]
images=[]
#loop through each item and extract details
for item in items:
  name=item.find('div',class_='name').text.strip()
  price=item.find('div',class_='price').text.strip()
  image_tag=item.find('div',class_='image').find('img')
  image_src=img_tag['src'] if image_tag else None #Extract src attributes

  #Append data to lists
  names.append(name)
  prices.append(price)
  images.append(image_src)

#print results
print("Names:",names)
print("prices:",prices)
print("Images:",images)
```

```python
df=pd.DataFrame({"Product_Name:":names,"MRP":prices,"Image_SRC:":images})
```
